```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

from sklearn.metrics import accuracy_score, classification_report
```

```python
dataset = pd.read_csv('news.csv')
```

```python
dataset
```

```python
independent = dataset['text']
dependent = dataset['label']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(independent, dependent, test_size = 0.2, random_state = 10)
```

```python
vectorizer = CountVectorizer(stop_words='english')

count_x_train = vectorizer.fit_transform(x_train)
count_x_test = vectorizer.transform(x_test)
```

```python
models = {
    "Naive Bayes": MultinomialNB(),
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=10),
    "Linear SVC": LinearSVC(random_state=10)
}

results = []

for name, model in models.items():
  
    model.fit(count_x_train, y_train)
    

    y_pred = model.predict(count_x_test)
    
    acc = accuracy_score(y_test, y_pred)
    results.append({"Model": name, "Accuracy": acc})
    
    print(f" {name} Accuracy: {acc:.4f}")

summary_df = pd.DataFrame(results).sort_values(by="Accuracy", ascending=False).reset_index(drop=True)
```

```python
# Display final ranking table
print("\nMODEL COMPARISON SUMMARY")
display(summary_df)

# Show best model
best_model_name = summary_df.loc[0, "Model"]
best_acc = summary_df.loc[0, "Accuracy"]
print(f"\n Best Performing Model: {best_model_name} with an Accuracy of {best_acc:.4f}")
```
