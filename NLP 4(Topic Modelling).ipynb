```python
import pandas as pd
```

```python
dataset = pd.read_csv("mabel.txt", header=None, on_bad_lines='skip', encoding='utf-8')
```

```python
dataset = dataset.drop(0)
```

```python
dataset.columns = ["Date", "Chat"]
```

```python
Message = dataset["Chat"].str.split("-", n=1, expand = True)
```

```python
dataset["Time"] = Message[0]
```

```python
Message2 = Message[1].str.split(":", n=1, expand = True)
```

```python
dataset["Name"] = Message2[0]
dataset["Message"] = Message2[1]
```

```python
dataset = dataset.drop(columns ="Chat")
```

```python
dataset
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
msg = dataset['Message'].dropna().values
```

```python
tf_vectorizer = CountVectorizer(stop_words='english')
tf = tf_vectorizer.fit_transform(msg)
```

```python
from sklearn.decomposition import LatentDirichletAllocation
```

```python
lda = LatentDirichletAllocation(n_components= 10, random_state=10)
lda.fit(tf)
```

```python
no_top_words = 5
feature_names = tf_vectorizer.get_feature_names_out()

for topic_idx, topic in enumerate(lda.components_):
    top_word_indices = topic.argsort()[:-no_top_words - 1:-1]
    
    top_words = [feature_names[i] for i in top_word_indices]
    
    print(f"Topic {topic_idx + 1}: " + ", ".join(top_words))
```
